> ⚠️ **This notebook is not pre-executed.** It calls the real Anthropic API and needs an `ANTHROPIC_API_KEY` environment variable set before you run it (get one at https://console.anthropic.com). Install dependencies first: `pip install -r requirements.txt`.

This notebook generates text with a frontier model (Claude) and reasons about the quality gap versus the local GPT-2 model from the previous notebook.

```python
import os

import anthropic

MODEL = "claude-opus-4-8"

assert os.environ.get("ANTHROPIC_API_KEY"), (
    "Set the ANTHROPIC_API_KEY environment variable before running this notebook. "
    "macOS/Linux: export ANTHROPIC_API_KEY='your-key-here'  "
    "Windows PowerShell: $env:ANTHROPIC_API_KEY = 'your-key-here'"
)

client = anthropic.Anthropic()
```

Same underlying mechanism as the previous two notebooks — predict a distribution over the next token and sample — but running on a frontier-scale model with training data, alignment, and safety work behind it that a hobby project could never replicate.

```python
def generate(client: anthropic.Anthropic, prompt: str) -> str:
    """Send one prompt to Claude and return the text of the reply."""
    response = client.messages.create(
        model=MODEL,
        max_tokens=200,
        messages=[{"role": "user", "content": prompt}],
    )
    return next(block.text for block in response.content if block.type == "text")
```

```python
prompt = "The future of artificial intelligence is"

print(f"Prompt: {prompt!r}\n")
print(f"Asking {MODEL} to complete this exact prompt used in the previous notebook:\n")

completion = generate(client, prompt)
print(f"  Claude's completion:\n  {completion}")
```

Now a task GPT-2 cannot meaningfully do: follow an instruction and reason about a real-world question.

```python
instruction_prompt = (
    "In exactly 3 bullet points, explain why a 124-million-parameter "
    "model like GPT-2 produces less coherent and less factually reliable "
    "text than a modern frontier LLM. Be concise."
)
print(f"Prompt: {instruction_prompt!r}\n")
answer = generate(client, instruction_prompt)
print(f"  Claude's answer:\n  {answer}")
```

## Why this comparison matters

Both GPT-2 and Claude do the same basic thing mechanically: predict a probability distribution over the next token and sample from it, repeatedly. The dramatic quality difference you just saw comes from three factors that scale independently of the core mechanism:

1. **Model size (parameters)** — GPT-2 small: 124M. Frontier models: orders of magnitude larger.
2. **Training data quality and volume** — far larger, more curated, more recent training corpora.
3. **Post-training alignment** — techniques like RLHF/instruction tuning that teach a model to actually follow instructions and be helpful/honest/harmless, rather than just continue text plausibly.
